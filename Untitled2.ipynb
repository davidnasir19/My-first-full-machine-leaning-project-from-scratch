```python
df = data.copy()
df.head()
```

```python
# creation des sous ensembles
```

```python
missing_rates = df.isna().sum()/df.shape[0]
blood_columns = df.columns[(missing_rates<0.90) & (missing_rates>0.88)]
viral_columns = df.columns[(missing_rates<0.88)&(missing_rates>0.75)]
keys_columns = ["Patient age quantile", "SARS-Cov-2 exam result"]
df = df[blood_columns.tolist()+viral_columns.tolist()+keys_columns]
df.head()
```

```python
# Train_test_split
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_set, test_set = train_test_split(df,test_size = 0.2, random_state = 0)
```

```python
train_set["SARS-Cov-2 exam result"].value_counts()
```

```python
test_set["SARS-Cov-2 exam result"].value_counts()
```

```python
# Encodage
```

```python
code = {
    "positive":1,
    "negative":0,
    "detected":1,
    "not_detected":0}
```

```python
for col in df.select_dtypes(include = ["object","string"]):
    df[col] = df[col].map(code).astype("float")
```

```python
df.head()
```

```python
def encodage(df):
    code = {
    "positive":1,
    "negative":0,
    "detected":1,
    "not_detected":0}
    for col in df.select_dtypes(include = ["object","string"]):
        df[col] = df[col].map(code).astype("float")
    return df
```

```python
def imputation(df):
    return df.dropna(axis = 0)
```

```python
def features_engenering(df):
    df["est malade"] = df[viral_columns].sum(axis = 1)>=1
    df = df.drop(columns=viral_columns)
    return df
```

```python
def preprocessing(df):
    df = encodage(df)
    df = features_engenering(df)
    df = imputation(df)
    x = df.drop("SARS-Cov-2 exam result", axis = 1)
    y = df["SARS-Cov-2 exam result"]
    print(y.value_counts())
    return x, y
```

```python
x_train, y_train = preprocessing(train_set)
```

```python
x_test, y_test = preprocessing(test_set)
```

```python
# Modelisation
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_selection import SelectKBest,f_classif
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import f1_score, confusion_matrix, classification_report
from sklearn.model_selection import learning_curve
```

```python
# model = DecisionTreeClassifier(random_state = 0)
preprocessor = make_pipeline(PolynomialFeatures(2, include_bias = False),SelectKBest(f_classif, k = 10))
```

```python
RandomForest = make_pipeline(preprocessor, RandomForestClassifier(random_state = 0))
AdaBoost = make_pipeline(preprocessor,AdaBoostClassifier(random_state = 0))
svm = make_pipeline(preprocessor,StandardScaler(),SVC(random_state = 0))
KNN = make_pipeline(preprocessor,StandardScaler(),KNeighborsClassifier())
```

```python
def evaluation(model):
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    n,train_score,val_score = learning_curve(model,x_train,y_train,train_sizes = np.linspace(0.1,1,10),cv = 4, scoring = "f1_macro")
    plt.figure(figsize=(12,8))
    plt.plot(n,train_score.mean(axis = 1),label = "train score")
    plt.plot(n,val_score.mean(axis = 1), label = "validation score")
    plt.legend()
```

```python
# listes des modeles de ml
```

```python
dict_of_models = {"RandomForest" :RandomForest,
                  "AdaBoost" : AdaBoost,
                    "svm" : svm,
                   "KNN" : KNN }
```

```python
# Procedure d'evaluation
```

```python
for name, model in dict_of_models.items():
    print(name)
    evaluation(model)
```

```python
# OPTIMISATION
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
svm
```

```python
hyper_params = {
    "svc__gamma": [1e-3, 1e-4],
    "svc__C": [1, 10, 100, 1000],
    "pipeline__polynomialfeatures__degree": [2, 3, 4],
    "pipeline__selectkbest__k": range(40, 60)
}
```

```python
grid = RandomizedSearchCV(svm,hyper_params, scoring = "recall",cv = 4, n_iter = 40)
grid.fit(x_train,y_train)
print(grid.best_params_)
y_pred = grid.predict(x_test)
print(classification_report(y_test,y_pred))
```

```python
evaluation(grid.best_estimator_)
```
